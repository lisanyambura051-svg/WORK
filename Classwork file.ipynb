```python
#Creating a dataset to demonstrate groupby
import pandas as pd
data = {
    "Name": ["John", "Jane","Doe", "Smith", "Jazmin"],
    "Age": [25, 30, 35, 40, 45],
    "Major": ["Computer","Science","engineering","mathematics","physics"],
    "Hobbies":["reading","sports","music","Travelling","swimming"],
    "GPA":[3.5, 3.8, 3.2, 3.9, 3.6  ],
    "Gender":["Male","Female","Male","Male","Female"],
    "Height" : [1.75,1.65,1.80,1.70,1.60]

}
df= pd.DataFrame(data)

#group by major and calculate mean GPA
grouped = df.groupby("Major")["GPA"].mean()
print(grouped)
```

```python
#aggregations with multiple functions
import  pandas as pd
data = {"Name": ["John", "Jane","Doe", "Smith", "Jazmin"],
    "Age": [25, 30, 35, 40, 45],
    "Major": ["Computer","Science","engineering","mathematics","physics"],
    "Hobbies":["reading","sports","music","Travelling","swimming"],
    "Gender":["Male","Female","Male","Male","Female"],
    "Height" : [1.75,1.65,1.80,1.70,1.60],
     "GPA":[3.5, 3.8, 3.2, 3.9, 3.6  ]
}
df = pd.DataFrame(data)
grouped = df.groupby("Major")["GPA"].agg(["mean", "min", "max"])
print(grouped)
```

```python
#Scales
#Nominal scale
import csv
import pandas as pd
df=pd.DataFrame(['A+','A-','B-','B+','C-','C+'],index=['excellent','excellent','good','good','ok','ok'])
df.rename(columns={0:'grades'}, inplace=True)
a = df['grades'].astype('category').head()
print(a)
```

```python
#Ordinal scale
import csv
import pandas as pd
df=pd.DataFrame(['A+','A-','B-','B+','C-','C+'],index=['excellent','excellent','good','good','ok','ok'])
df.rename(columns={0:'grades'}, inplace=True)
grades = pd.Categorical(df['grades'], categories=['A+','A-','B-','B+','C-','C+'], ordered=True)
print(grades)
```
